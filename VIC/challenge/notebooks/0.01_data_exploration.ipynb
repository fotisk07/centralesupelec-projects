```python
import matplotlib.pyplot as plt
import numpy as np

from challenge.dataset import Dataset
from challenge.visuals import plot_image_with_boxes
```

```python
data = Dataset("../data/data.yaml", "train")
```

```python
idx = np.random.randint(0, len(data), 6)

fig, ax = plt.subplots(2, 3, figsize=(12, 8))
ax = ax.flatten()

for i, j in enumerate(idx):
    plot_image_with_boxes(
        ax[i],
        data[j]["image"],
        data[j]["targets"],
        color="red",
    )

plt.tight_layout()
plt.show()
```

# Exploratory Data analysis

## Intro

```python
objs = []
for sample in data:
    objs.extend(sample["targets"])
```

```python
from collections import Counter

id_counts = Counter([int(obj[0]) for obj in objs])
obj_counts = {data.metadata["names"][k]: v for k, v in id_counts.items()}
```

```python
print("Dataset size", len(data))
print("Number of objects", len(objs))
print("Counts", obj_counts)
```

## Objects per image

```python
obj_per_img = []
for sample in data:
    obj_per_img.append(len(sample["targets"]))
```

```python
import seaborn as sns

sns.kdeplot(obj_per_img)
```
